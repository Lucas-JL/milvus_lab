Milvus 资源组关键说明
Milvus 默认有 __default_resource_group__，未指定的集合会绑定到该组；
资源组的 “capacity” 不是硬限制，而是 Milvus 调度 QueryNode 资源的参考值；
仅 QueryNode 支持资源组，DataNode/IndexNode 暂不支持（Milvus 2.4+ 版本）。

磁盘配额是否包含副本？  
Milvus 的磁盘配额不直接配置 “数据库级” 参数，而是通过以下方式管控，且副本会占用额外磁盘：
Milvus 的数据存储（如 MinIO/S3、本地磁盘）是按实例级管控，而非数据库级；
集合的副本（Replica）会存储相同数据，因此总磁盘占用 = 单副本数据量 × 副本数（与你之前问的 “包含副本” 逻辑一致）。
